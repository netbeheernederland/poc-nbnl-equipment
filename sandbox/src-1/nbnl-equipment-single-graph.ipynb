```python
import json
import os
import shutil
from pathlib import Path

import yaml
from rdflib import Graph
from pyshacl import validate


SRCDIR = Path(".")
BUILDDIR = SRCDIR / ".." / "build"
DATADIR = SRCDIR / ".." / "data"


def yaml2json(y):
    os.makedirs(BUILDDIR / y.parent, exist_ok=True)
    with y.open("rb") as f:
        data = yaml.safe_load(f)
    with open(BUILDDIR / y.with_suffix(".jsonld"), "w") as g:
        json.dump(data, g, default=str)


shutil.rmtree(BUILDDIR, ignore_errors=True)
yaml2json(SRCDIR / "nbnl-equipment.shacl.yamlld")

data = Graph()
data.parse(DATADIR / "AS_EQ_OP_Westland_FGU" / "equipment_four_stations_EQ.jsonld", format="json-ld")

conforms, results_graph, results_text = validate(
    data,
    shacl_graph=BUILDDIR / "nbnl-equipment.shacl.jsonld",
    data_graph_format=RdfFormat.JSON_LD,
    shacl_graph_format=RdfFormat.JSON_LD,
    advanced=True,
)

print(results_text)
```
